#📘 Guia Rápido de Prompting para Engenharia (Cheat Sheet)
* **Defina o Papel:** Comece sempre com "Atue como um Engenheiro de [Área]..."

* **Dê o Contexto Físico:** Especifique o hardware (ESP32, PLC Siemens S7-1200, Arduino Uno).

* **Estabeleça Restrições:** "O código deve ser não-bloqueante (sem usar delay())", "Siga a norma MISRA C".

* **Peça Explicação:** "Explique cada linha do código focado na segurança do operador".

* **Iteração:** Se a IA errar, não comece do zero. Diga: "O erro X ocorreu na linha Y, corrija considerando Z".
